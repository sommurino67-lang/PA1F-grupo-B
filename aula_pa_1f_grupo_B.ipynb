```python
#projeto caixa eletronico

#voce foi contratado para desenvolver um sistema simples de caixa eletronico

#o sistema deverá funcionar através de um menu interativo usando "while"

#o programa deve iniciar com um saldo inicial de R$1000,00 e permitir que o usuario escolha diferentes operações até decidir sair do sistema

# ver valor
#depositar
#sacar
#sair

#regras

#quando depositar somar ao saldo anterior;
#quando sacar subtrair do saldo anterior
#mostrar o saldo para o usuario

saldo = 1000.00
opacao = 0

print("=========== MENU ==========")
print("")
print("opção 1 - Depositar")
print("opção 2 - Sacar")
print("opção 3 - Ver Saldo")
print("opção 4 - Sair")
print("===========================")
print("")

while opacao != 4:
  try:
    opacao = int(input("Digite a opção desejada ou 4 para sair!:   "))

    if opacao == 4:
      print("saindo do caixa")
      break
    elif opacao == 1:
      valor_deposito = float(input("Digite o valor para depositar: "))
      if valor_deposito > 0:
        saldo += valor_deposito
        print(f"Depósito de R${valor_deposito:.2f} realizado com sucesso. Novo saldo: R${saldo:.2f}")
      print("")
    elif opacao == 2:
      valor_saque = float(input("Digite o valor para sacar: "))
      if valor_saque > 0:
        if saldo >= valor_saque:
          saldo -= valor_saque
          print(f"Saque de R${valor_saque:.2f} realizado com sucesso. Novo saldo: R${saldo:.2f}")
      print("")
    elif opacao == 3:
      print(f"Seu saldo atual é: R${saldo:.2f}")
      print("")

  except :
    print("Opção inválida. Por favor, digite um número entre 1 e 4.")
```

```python
import random

# Gerar número secreto
numero_secreto = random.randint(0, 10)
tentativas = 3

print("=== Jogo de Adivinhação ===")
print("Você tem 3 chances para acertar o número de 0 a 10!")

while tentativas > 0:
    try:
        chute = int(input(f"\nTentativas restantes: {tentativas}. Digite seu palpite: "))

        if chute == numero_secreto:
            print("Parabéns! Você acertou!")
            break

        # Lógica de proximidade
        distancia = abs(chute - numero_secreto)

        if distancia <= 1:
            print("Está QUENTE!")
        elif distancia <= 3:
            print("Está MORNO.")
        else:
            print("Está FRIO.")

        if chute < numero_secreto:
            print("Dica: O número é MAIOR.")
        else:
            print("Dica: O número é MENOR.")

        tentativas -= 1

    except ValueError:
        print("Por favor, digite um número válido.")

if tentativas == 0:
    print(f"\nFim de jogo! O número era {numero_secreto}.")
```

### Questão 01 - Contagem Crescente
Este programa exibe os números de 1 a 20.

```python
for i in range(1, 21):
    print(i)
```

```python
n = int(input("Digite um número inteiro positivo: "))
soma = 0

for i in range(1, n + 1):
    soma += i

print(f"Resultado: {soma}")
```

```python
num = int(input("Digite um número: "))

for i in range(1, 11):
    resultado = num * i
    print(f"{num} x {i} = {resultado}")
```

```python
num = int(input("Digite um número: "))

if num < 0:
    print("Por favor, digite um número positivo.")
else:
    fatorial = 1
    for i in range(1, num + 1):
        fatorial *= i
    print(f"Fatorial = {fatorial}")
```

```python
valor = int(input("Digite o valor do saque: "))

# Definimos as notas disponíveis em ordem decrescente
# Como não podemos usar listas, vamos iterar manualmente usando a lógica de notas

for nota in [100, 50, 20, 10, 5, 2, 1]:
    quantidade = valor // nota
    if quantidade > 0:
        print(f"{quantidade} nota(s) de R${nota}")
        valor %= nota
```
